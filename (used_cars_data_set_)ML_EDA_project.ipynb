# 4. Handling Outliers and Skewness

Outliers in car_price are identified and removed using the Interquartile Range (IQR) method

```python
def remove_outliers_iqr(df, column):
    # Calculate Q1 (25th percentile) and Q3 (75th percentile) for the column
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    # Calculate the Interquartile Range (IQR)
    IQR = Q3 - Q1

    # Define the lower and upper bounds for detecting outliers
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Filter out the outliers from the DataFrame
    df_filtered = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    return df_filtered


df_no_outliers = remove_outliers_iqr(df,'car_price')
```

# 5. Train-Test Split and Model Training

Feature Selection:

Independent features: ['car_engine_capacity', 'car_mileage', 'car_age', 'car_country', 'car_drive', 'car_engine_hp']
Dependent feature: ['car_price']

```python
independent_feature=['car_engine_capacity','car_mileage','car_age','car_country','car_drive','car_engine_hp']#independent feature
dependent_feature=['car_price']#dependent feature
x=df[independent_feature]
y=df[dependent_feature]# (y)target variable
```

The data is split into training (80%) and testing (20%) sets

Splitting the Data:

train_test_split divides data into training (80%) and testing (20%) sets with random_state=42 for reproducibility.

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,#0.2 means 20% of data specified
#for testing set,and the other 80% for training set
                                               random_state=42)# to take the same data for training and testing set


print(x_train.shape)  # Prints the shape of x_train (rows, columns)
print(x_test.shape)   # Prints the shape of x_test
print(y_train.shape)  # Prints the shape of y_train
print(y_test.shape)   # Prints the shape of y_test
```

```python
from sklearn.ensemble import RandomForestRegressor
```

Model Selection and Training:

A RandomForestRegressor model is used for training.
Model training: model.fit(x_train, y_train)


```python
model=RandomForestRegressor()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
y_pred
```

# 6. Model Evaluation

>Predictions on test data: y_pred = model.predict(x_test)
